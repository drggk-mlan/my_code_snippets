```python
class Timer:#@save

    def __init__(self):
        self.times=[]
        self.start()

    def start(self):
        self.tik=time.time()#开始时间

    def stop(self):
        self.times.append(time.time()-self.tik())
        return self.times[-1]

    def avg(self):
        return sum(self.times)/len(self.times)
    
    def sum(self):
        return sum(self.times)
    
    def cumsum(self):
        return np.array(self.times).cumsum().tolist()
```

```python
def normal(x,mu,sigma):
    p=1/math.sqrt(2*math.pi*sigma**2)
    return p*np.exp(-0.5/sigma**2*(x-mu)**2)
```

```python
x=np.arange(-7,7,0.01)

params=[(0,1),(0,2),(3,1)]

d2l.plot(x,[normal(x,mu,sigma) for mu,sigma in params],xlabel='x',ylabel='p(x)',figsize=(4.5,2.5),legend=['$\mu=0,\sigma=1$','$\mu=0,\sigma=2$','$\mu=3,\sigma=1$'])
```
